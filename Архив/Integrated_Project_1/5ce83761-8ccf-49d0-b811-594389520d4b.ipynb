- Заглушка везде мешает, по этому при анализе целевого параметра, который в датасете содержал NaN, всегда отфильтровываем ее. Заглушку лучше ставить далеко вне диапазона целевых параметров. Если вдруг забудем ее отфильтровать, то сразу заметим при преобразованиях. Кроме того, заглушка показывает, что оценили количество пропусков в датасете и приняли по ним обоснованное решение. Необходимо помнить, что не отфильтрованная заглушка в числовых данных будет искажать расчеты средних, медиан и диаграммы размаха. <br>Но можем сделать анализ и с учетом заглушки и обнаружить неявный признак, который нес пропуск.</font>

```python
# Фильтруем фильмы с государственной поддержкой
df_supported = df[df['has_support']]

# Убедимся, что у этих фильмов есть данные о бюджете и сборах
df_supported = df_supported[(df_supported['budget'] > 0) & (df_supported['box_office'] > 0)]
```

```python
# Фильтруем фильмы без государственной поддержки
df_unsupported = df[~df['has_support']]

# Убедимся, что у этих фильмов есть данные о бюджете и сборах
df_unsupported = df_unsupported[(df_unsupported['budget'] > 0) & (df_unsupported['box_office'] > 0)]

# Рассчитываем окупаемость
df_unsupported['profitability'] = df_unsupported['box_office'] / df_unsupported['budget']

# Средняя окупаемость для фильмов без поддержки
mean_profitability_unsupported = df_unsupported['profitability'].mean()

# Доля окупившихся фильмов без поддержки
profitable_ratio_unsupported = (df_unsupported['profitability'] > 1).mean()

print("\nСравнение с фильмами без государственной поддержки:")
print(f"- Средняя окупаемость: {mean_profitability_unsupported:.2f}")
print(f"- Доля окупившихся фильмов: {profitable_ratio_unsupported:.2%}")

print("Анализ окупаемости фильмов с государственной поддержкой:")
print(f"- Средняя окупаемость: {mean_profitability:.2f}")
print(f"- Доля окупившихся фильмов: {profitable_ratio:.2%}")
```
