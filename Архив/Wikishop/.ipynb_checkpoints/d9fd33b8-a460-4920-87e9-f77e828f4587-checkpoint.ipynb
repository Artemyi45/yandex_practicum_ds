# Проект для "Викишоп"

## Описание проекта:
Интернет-магазин «Викишоп» запускает новый сервис. Теперь пользователи могут редактировать и дополнять описания товаров, как в вики-сообществах. То есть клиенты предлагают свои правки и комментируют изменения других. Магазину нужен инструмент, который будет искать токсичные комментарии и отправлять их на модерацию. 

## Цель проекта:
Магазину нужен инструмент, который будет искать токсичные комментарии и отправлять их на модерацию.  
- Обучить модель классифицировать комментарии на позитивные и негативные.
- Построить модель со значением метрики качества F1 не меньше 0.75. 


## План проекта:
**1.** Загрузка и первичный анализ данных  
**2.** Предобработка  
**3.** Разделение данных train/test  
**4.** Векторизация текста  
**5.** Обучение и тестирование моделей  
**6.** Итоговый вывод  

## Описание данных:
Датасет **toxic_comments**  
Столбец **text** в нём содержит текст комментария, а **toxic** — целевой признак.

### 2. Предобработка данных

Уберем неинформативный столбец "Unnamed: 0". Так же уберем шум из текста. Иными словами приведем текст к "чистому" виду.

```python
# удаляем столбец

data = data.drop(columns = 'Unnamed: 0')
```

<img src="https://emojigraph.org/media/apple/check-mark-button_2705.png" align=left width=33, heigth=33>
<div class="alert alert-success">
    
    
Да, все верно, столбец `'Unnamed: 0'` можно безболезненно удалить. Это "старый" индекс, который похоже забыли при пересохранении датасета
    
------    
    
 Давай поймем что это за волшебная колонка  `Unnamed: 0` и откуда она взялась.

Когда ты сохраняешь `DataFrame` с помощью `.to_csv()`, Pandas по умолчанию добавляет индекс (нумерацию строк) в файл.
    
А при загрузке этого файла через `pd.read_csv()` без дополнительных параметров, этот индекс читается как обычная колонка с названием `Unnamed: 0`.
    
Чтобы этого избежать добавь `index=False` при сохранении либо `index_col=0` при чтении.
    
    

</div>

```python
# проверка

data.columns
```

```python
# функция для очистки текста

def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^a-zа-яё\s]', '', text)
    text = re.sub(r'\s+', ' ', text)
    text = text.strip()
    
    return text
```

```python
# применяем функцию ко всем строкам в колонке 'text''

data['text_clean'] = data['text'].apply(clean_text)
```

</div>

```python
# проверка

print(f'Исходный: {data["text"].iloc[0]}')
print('--' * 50)
print(f'Очищенный: {data["text_clean"].iloc[0]}')
```

### 3. Разделение данных на train/val/test

Разобьем данные на тренировочную, валидационную и тестовую выборки.

```python
# делаем тестовую выборку

X_temp, X_test, y_temp, y_test = train_test_split(
    data['text_clean'], data['toxic'],
    test_size = 0.15,
    random_state = 42,
    stratify = data['toxic']
)

X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp,
    test_size = 0.176,
    random_state = 42,
    stratify = y_temp
)
```

```python
# проверяем размеры

print(f"Train: {len(X_train)} ({len(X_train)/len(data)*100:.1f}%)")
print(f"Val: {len(X_val)} ({len(X_val)/len(data)*100:.1f}%)")
print(f"Test: {len(X_test)} ({len(X_test)/len(data)*100:.1f}%)")
```

```python
# проверяем пропорции классов

print("Доля toxic в train:", y_train.mean().round(3))
print("Доля toxic в val:", y_val.mean().round(3))
print("Доля toxic в test:", y_test.mean().round(3))
```

<img src="https://emojigraph.org/media/apple/check-mark-button_2705.png" align=left width=33, heigth=33>
<div class="alert alert-success">
    
Данные разделены на выборки. Но я бы посоветовал тебе изменить пропорции и выделить на тест 10%. Причина простая - модели, которые мы обучаем чувствительны к объему обучающих данных. Чем больше слов они увидят в процессе обучения, и оценят их вклад в "токсичность", тем лучше будут модели. А для корректного тестирования и 10% данных вполне достаточно, учитывая немаленький размер датасета.

</div>

**Вывод по разделу 3:**  
Разбили данные на тренировочную, валидационную и тестовую выборки. Сохранили пропорцию классов.

### 4. Векторизация текста

Превратим текст в числа, с помощью TF - IDF.

```python
# создаем векторизатор и проводим обучение и преобразование

vectorizer = TfidfVectorizer(
    ngram_range=(1, 2),
    max_features=10000,
    sublinear_tf=True
)

X_train_tfidf = vectorizer.fit_transform(X_train)

X_val_tfidf = vectorizer.transform(X_val)
X_test_tfidf = vectorizer.transform(X_test)
```

<img src="https://emojigraph.org/media/apple/check-mark-button_2705.png" align=left width=33, heigth=33>
<div class="alert alert-success">
👍 Молодец, получены признаки из текстов с помощью TF-IDF. И совершенно верно векторайзер вначале обучен на обучающем наборе и затем применен к валидационному и тестовому</div>

<img src="https://upload.wikimedia.org/wikipedia/commons/b/ba/Warning_sign_4.0.png" align=left width=44, heigth=33>
<div class="alert alert-warning">
Имей в виду - размерность вектора признаков это главный trade-off (компромис). В одной стороны, чем больше признаков, тем лучше потенциально могут быть метрики, т.к. учитывается больше нюансов.  С другой стороны мы расплачиваемя за это временем обучения.

10_000 признаков в этом проекте ближе к  минимальной границе. Моя интуитивная оценка оптимума - это около 50_000. Поэтому если метрики низкие - попробуй покрутить настройки именно здесь.
</div>

```python
# проверка

print("Размер матрицы train (строки, столбцы):", X_train_tfidf.shape)
print("Размер матрицы val:", X_val_tfidf.shape)
print("Размер матрицы test:", X_test_tfidf.shape)
```

**Вывод по разделу 4:**  
Сделали векторизация текста, с помощью TF - IDF.

### 5. Обучение и тестирование моделей

Сделаем первую модель - логистическую регрессию. Будем использовать ее результаты как baseline.

```python
# Linear Regression

model = LogisticRegression(
    class_weight='balanced',
    max_iter=1000,
    random_state=42
)

model.fit(X_train_tfidf, y_train)

y_val_pred = model.predict(X_val_tfidf)
```

```python
# проверка на валидации

f1 = f1_score(y_val, y_val_pred)
print(f"F1 на валидации: {f1:.4f}")
```

Далее попробуем подобрать гиперпараметры.

```python
%%time

# пайплайн, сетка гиперпараметров и обучение


pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(sublinear_tf=True)),
    ('clf', LogisticRegression(class_weight='balanced', max_iter=1000))
])

param_grid = {
    'tfidf__ngram_range': [(1, 1), (1, 2)],
    'tfidf__max_features': [5000, 10000, 15000],
    'clf__C': [0.1, 1.0, 10.0]
}

grid = GridSearchCV(
    pipeline,
    param_grid,
    cv=3,
    scoring='f1',
    n_jobs=-1
)

grid.fit(X_train, y_train)

print("Лучшие параметры:", grid.best_params_)
print("Лучший F1 на кросс-валидации:", grid.best_score_)
```

Поэтому я бы советовал выбрать какую-то одну стратегию (отдельная валидационная выборка или кросс-валидация).

```python
# оценка на тесте

y_test_pred = grid.predict(X_test)
print("F1 на тесте:", f1_score(y_test, y_test_pred))
```

<img src="https://emojigraph.org/media/apple/check-mark-button_2705.png" align=left width=33, heigth=33>
<div class="alert alert-success">
Молодец: 
    
  - выбрана лучшая модель
  - тестирование произведено корректно 
  - достигнута метрика f1 выше 0,75 на тестовой выборке
</div>


```python
# random forest

param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [10, 20, None],
    'min_samples_split': [2, 5, 10],
    'class_weight': ['balanced', 'balanced_subsample']
}

rf_random = RandomizedSearchCV(
    RandomForestClassifier(random_state=42, n_jobs=-1),
    param_grid,
    n_iter=5,
    cv=2,
    scoring='f1',
    random_state=42,
    n_jobs=-1,
    verbose=1
)

rf_random.fit(X_train_tfidf, y_train)

print("Лучшие параметры:", rf_random.best_params_)
print("Лучший F1 на CV:", rf_random.best_score_)

best_rf = rf_random.best_estimator_
y_pred_rf_tuned = best_rf.predict(X_test_tfidf)
f1_rf_tuned = f1_score(y_test, y_pred_rf_tuned)

print(f"F1 на тесте (Random Forest tuned): {f1_rf_tuned:.4f}")
```

Обучили модель Random Forest. Получили результат 0.62 на метрике f1.
