## Описание данных

1. Файл ferma_main.csv содержит данные о стаде фермера на текущий момент. Описание данных:
    - id — уникальный идентификатор коровы.
    - Удой, кг — масса молока, которую корова даёт в год (в килограммах).
    - ЭКЕ (Энергетическая кормовая единица) — показатель питательности корма коровы.
    - Сырой протеин, г — содержание сырого протеина в корме (в граммах).
    - СПО (Сахаро-протеиновое соотношение) — отношение сахара к протеину в корме коровы.
    - Порода — порода коровы.
    - Тип пастбища — ландшафт лугов, на которых паслась корова.
    - Порода папы_быка — порода папы коровы.
    - Жирность, % — содержание жиров в молоке (в процентах).
    - Белок, % — содержание белков в молоке (в процентах).
    - Вкус молока — оценка вкуса по личным критериям фермера, бинарный признак (вкусно, не вкусно).
    - Возраст — возраст коровы, бинарный признак (менее_2_лет, более_2_лет).
 - Признаки в датасете ferma_main.csv можно разделить на группы.
    - Характеристики коровы: id, Порода, порода папы_быка, Возраст.
    - Характеристики корма: ЭКЕ (Энергетическая кормовая единица), Сырой протеин, г, СПО (Сахаро-протеиновое соотношение).
    - Характеристика пастбища: Тип пастбища.
    - Характеристики молока: Удой, кг, Жирность,%, Белок,%, Вкус молока.
 - Такое разделение поможет вам понять, как взаимосвязаны признаки в датасете.


2. Файл ferma_dad.csv хранит имя папы каждой коровы в стаде фермера. Описание данных:
    - id — уникальный идентификатор коровы.
    - Имя Папы — имя папы коровы.


3. Файл cow_buy.csv — это данные о коровах «ЭкоФермы», которых фермер хочет изучить перед покупкой. Описание данных:
    - Порода — порода коровы.
    - Тип пастбища — ландшафт лугов, на которых паслась корова.
    - Порода папы_быка — порода папы коровы.
    - Имя_папы — имя папы коровы.
    - Текущая_жирность,% — содержание жиров в молоке (в процентах).
    - Текущий_уровень_белок,% — содержание белков в молоке (в процентах).
    - Возраст — возраст коровы, бинарный признак (менее_2_лет, более_2_лет).
 - Данные в файлах ferma_main.csv и cow_buy.csv похожи, но у датасета cow_buy.csv несколько особенностей.
    - Содержание белков и жиров в молоке указано на момент продажи — когда «ЭкоФерма» давала коровам свой корм.
    - Параметры корма ЭКЕ (Энергетическая кормовая единица), Сырой протеин, г и СПО (Сахаро-протеиновое соотношение) отсутствуют. Технологи заказчика пересмотрели подход к кормлению: для новых коров планируется увеличить значения каждого из этих параметров на 5%.
    - Кроме того, отсутствуют признаки Удой, кг и Вкус молока. Это целевые признаки, именно их предстоит предсказать.

```python
low = ferma_main['Fat'].quantile(0.33)
medium = ferma_main['Fat'].quantile(0.66)

print(f'''
Границы категорий:
- Низкая: < {low:.3f}%
- Средняя: {low:.3f}% — {medium:.3f}%
- Высокая: > {medium:.3f}%
''')
```

```python
def categorize_fat(value):
    if value < low:
        return 'low'
    elif value < medium:
        return 'medium'
    else:
        return 'high'

ferma_main['Fat'] = ferma_main['Fat'].apply(categorize_fat)
```

```python
model_one = LinearRegression()

X_one = ferma_clean.drop('Milk', axis = 1)
y_one = ferma_clean['Milk']

X_train_one, X_test_one, y_train_one, y_test_one = train_test_split(
    X_one, y_one,
    test_size = 0.25,
    random_state = 42
)

num_cols_one = [col for col in X_one.select_dtypes('number').columns if col != 'Milk']
cat_cols_one = X_one.select_dtypes('object').columns.tolist()

encoder_one = OneHotEncoder(handle_unknown='ignore')
encoder_one.fit(X_train_one[cat_cols_one])
X_train_one_ohe = encoder_one.transform(X_train_one[cat_cols_one])
X_test_one_ohe = encoder_one.transform(X_test_one[cat_cols_one])

scaler_one = StandardScaler()
scaler_one.fit(X_train_one[num_cols_one])
X_train_one_sca = scaler_one.transform(X_train_one[num_cols_one])
X_test_one_sca = scaler_one.transform(X_test_one[num_cols_one])

X_train_one = hstack([X_train_one_ohe, X_train_one_sca])
X_test_one = hstack([X_test_one_ohe, X_test_one_sca])

model_one.fit(X_train_one, y_train_one)

predictions_one = model_one.predict(X_test_one)
print('R2 первой модели:', r2_score(y_test_one, predictions_one))
```

```python
def calculate_metrics(y_true, y_pred, model_name):
    return {
        'Модель': model_name,
        'R²': r2_score(y_true, y_pred),
        'MAE': mean_absolute_error(y_true, y_pred),
        'RMSE': np.sqrt(mean_squared_error(y_true, y_pred))
    }

metrics = [
    calculate_metrics(y_test_one, predictions_one, "Модель 1 (базовая)"),
    calculate_metrics(y_test_two, predictions_two, "Модель 2 (SPO_2 + Energy²)"), 
    calculate_metrics(y_test_three, predictions_three, "Модель 3 (с данными о папе)")
]

results = pd.DataFrame(metrics).set_index('Модель')

def highlight_max(s):
    is_max = s == s.max()
    return ['background-color: #98FB98' if v else '' for v in is_max]

styled_results = results.style \
    .format({'R²': '{:.3f}', 'MAE': '{:.1f}', 'RMSE': '{:.1f}'}) \
    .apply(highlight_max, subset=['R²']) \
    .set_caption('Сравнение моделей линейной регрессии') \
    .set_properties(**{'text-align': 'center'})

display(styled_results)
```

```python
X = ferma_clean_class.drop(columns = ['Flavor'])
y = ferma_clean_class['Flavor']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size = 0.25,
    random_state = 42
)

encoder_class = OneHotEncoder(handle_unknown='ignore')
scaler_class = StandardScaler()

num_cols_class = ['SPO', 'Energy', 'Age']
cat_cols_class = ['Pasture', 'Fat']

encoder_class.fit(X_train[cat_cols_class])
X_train_ohe = encoder_class.transform(X_train[cat_cols_class])
X_test_ohe = encoder_class.transform(X_test[cat_cols_class])

scaler_class.fit(X_train[num_cols_class])
X_train_sca = scaler_class.transform(X_train[num_cols_class])
X_test_sca = scaler_class.transform(X_test[num_cols_class])

X_train_class = hstack([X_train_ohe, X_train_sca])
X_test_class = hstack([X_test_ohe, X_test_sca])

model_class = LogisticRegression()
model_class.fit(X_train_class, y_train)

y_pred = model_class.predict(X_test_class)
y_proba = model_class.predict_proba(X_test_class)
```

Попробуем свести ошибку 1 рода к нулю, для этого изменим значение порога. Цель увеличить Recall до максимума.

```python
y_proba_positive = y_proba[:, 1] if len(y_proba.shape) == 2 else y_proba

precision, recall, thresholds = precision_recall_curve(y_test, y_proba_positive)

plt.figure(figsize=(8, 6))
plt.plot(thresholds, precision[:-1], label="Precision")
plt.plot(thresholds, recall[:-1], label="Recall")
plt.xlabel("Порог классификации")
plt.ylabel("Значение метрики")
plt.title("Зависимость precision и recall от порога")
plt.legend()
plt.grid(True)
plt.show()

optimal_idx = np.argmax(precision)
optimal_threshold = thresholds[optimal_idx]
print(f"Оптимальный порог: {optimal_threshold:.2f}")
print(f"Максимальная precision: {precision[optimal_idx]:.2f}")
print(f"Recall при этом пороге: {recall[optimal_idx]:.2f}")
```
